```python
# Cell 1: Setup & Imports
import sys
import os
sys.path.append(os.path.abspath('..'))

from src.preprocessing import load_and_preprocess_data
from src.model import run_knn_sweep
from src.evaluation import summarize_results
```

```python
# Cell 2: Config
DATASET_LABEL = "Dataset A"
CSV_PATH = "../data/diabetes.csv"
TARGET_COLUMN = "Outcome"
PAPER_ACCURACY = 0.8571
K_VALUES = list(range(1, 31, 2))
MINKOWSKI_P = 3
```

```python
# Cell 3: Preprocessing & Training Pipeline
X_train, X_test, y_train, y_test = load_and_preprocess_data(
    csv_path=CSV_PATH, 
    target_col=TARGET_COLUMN
)

results_df = run_knn_sweep(
    X_train, X_test, y_train, y_test, 
    k_values=K_VALUES, 
    minkowski_p=MINKOWSKI_P
)
```

```python
# Cell 4: Evaluation Summary
best_metrics_df = summarize_results(
    results_df=results_df, 
    dataset_label=DATASET_LABEL, 
    paper_accuracy=PAPER_ACCURACY
)

display(best_metrics_df)
```
